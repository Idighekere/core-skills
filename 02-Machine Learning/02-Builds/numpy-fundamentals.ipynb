```python
import numpy as np
arr = np.random.randn(5, 3)
# compute the mean of each column, no loop
col_mean = arr.mean(axis=0)
print(col_mean)
# compute the mean of each row, no loop
row_mean = arr.mean(axis=1)
print(row_mean)
# normalize the array (subtract mean, divide by std)
col_std = arr.std(axis=0)
normalized = (arr - col_mean) / col_std
print("normalized array:", normalized)
```

some correction on the above.

Initially thought that computing the mean of each column will mean `axis=1` since 1 refers to column, but looking at it:
    - in order to get the mean of each column we need to compute the values across all the rows on each column, that makes `axis=0`
