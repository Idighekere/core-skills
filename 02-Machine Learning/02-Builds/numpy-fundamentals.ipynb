When creating arrays using the arange functions using floating point argument, the number of elements mght not be known due to some precision with  decimals, so it's prefereable to use `np.linspace`

```python
A = np.array([[1,1],
              [0,1]])

B = np.array([[2,0],
              [3,4]])

print( A*B )
print( A@B ) #this si the dot product of A and B
print(A.dot(B))
```
